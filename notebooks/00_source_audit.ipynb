# Source permissions and viability audit

Bounded public-source inspection completed **2026-10-31**. This notebook reads the committed `data/source_registry.csv` and factual `data/source_audit.json`; it makes **no network requests**, needs no credentials and can run offline in CI. Code-cell outputs are intentionally not fabricated or committed.

This is a source audit, not data acquisition, a coverage analysis, a predictor evaluation or a normalized precursor dataset. Publisher text, charts, biographies and images are not included. Registry retrieval times are the common UTC completion time of the inspection batch; source-version dates, event dates and historical prediction timestamps are different concepts.

```python
import csv
import json
from collections import Counter
from datetime import datetime
from pathlib import Path

cwd = Path.cwd().resolve()
root = next((p for p in (cwd, *cwd.parents) if (p / 'data/source_registry.csv').is_file()), None)
if root is None:
    raise FileNotFoundError('Run from the project root or a descendant; committed source registry missing.')
with (root / 'data/source_registry.csv').open(newline='', encoding='utf-8') as stream:
    reader = csv.DictReader(stream)
    columns = reader.fieldnames
    registry = list(reader)
audit = json.loads((root / 'data/source_audit.json').read_text(encoding='utf-8'))
evidence = {item['id']: item for item in audit['evidence']}
sources = {row['source_id']: row for row in registry}
print(f'{len(registry)} source entries; {len(evidence)} evidence records; audit {audit["audit_completed_at"]}')
```

```python
def show_table(rows, fields):
    print(' | '.join(fields))
    print(' | '.join('---' for _ in fields))
    for row in rows:
        print(' | '.join(str(row[field]).replace('\n', ' ') for field in fields))

show_table(registry, ['source_id', 'inspection_status', 'permission_decision', 'evidence_ids'])
print('\nSource permission decisions (not ballot completeness):', dict(Counter(r['permission_decision'] for r in registry)))
print('\nCore blockers:')
for blocker in audit['decision']['core_blockers']:
    print('-', blocker)
```

```python
# Validate the committed audit, not the underlying awards data or its legal clearance.
assert registry and len(sources) == len(registry), 'Source IDs must be unique.'
assert len(evidence) == len(audit['evidence']), 'Evidence IDs must be unique.'
assert len(columns) == len(set(columns)), 'Registry columns must be unique.'
for row in registry:
    assert None not in row and all(value is not None for value in row.values()), 'Malformed CSV row.'
    assert row['source_url'].startswith('https://')
    retrieved = datetime.fromisoformat(row['retrieved_at'].replace('Z', '+00:00'))
    assert retrieved.tzinfo is not None
    assert row['sha256'] == '', 'No acquisition snapshots were checksummed in this audit.'
    linked = row['evidence_ids'].split('|')
    assert all(key in evidence for key in linked), f'Unknown evidence for {row["source_id"]}'
    assert all(row['source_id'] in evidence[key]['source_ids'] for key in linked)
for item in evidence.values():
    assert all(key in sources for key in item['source_ids'])
assert audit['decision']['primary_selected'] is None
assert audit['decision']['fallback_selected'] is None
print('Registry structure, timezone-aware retrievals and evidence references validated; rights/coverage remain explicitly unresolved.')
```
